Eliminar registros con valores en pais nulos.

```python
df_h1.dropna(subset="Country", inplace= True)
df_h1
```

Rellenar valores nulos con meidas por pais.


```python
df_h1["ABV"] = df_h1["ABV"].str.replace("ABV ", "")
df_h1["ABV"] = df_h1["ABV"].str.replace("%", "").astype(float)

df_h1.info()
```
