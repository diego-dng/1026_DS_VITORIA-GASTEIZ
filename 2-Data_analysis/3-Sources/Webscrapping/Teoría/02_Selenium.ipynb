# Web Scraping: Selenium

A menudo, los datos están disponibles públicamente para nosotros, pero no en una forma que sea fácilmente utilizable. Ahí es donde entra en juego el web scraping, podemos usar web scraping para obtener nuestros datos deseados en un formato conveniente que luego se puede usar. A continuación, mostraré cómo se puede extraer información de interés de un sitio web usando el paquete Selenium en Python. Selenium nos permite manejar una ventana del navegador e interactuar con el sitio web mediante programación. 

Selenium también tiene varios métodos que facilitan la extracción de datos.

Antes de comenzar se preguntaran si ya se BeautifulSoup cual es la diferencia con Selenium.

A diferencia BeautifulSoup, Selenium no trabaja con el texto fuente en HTML de la web en cuestión, sino que carga la página en un navegador sin interfaz de usuario. El navegador interpreta entonces el código fuente de la página y crea, a partir de él, un Document Object Model (modelo de objetos de documento o DOM). Esta interfaz estandarizada permite poner a prueba las interacciones de los usuarios. De esta forma se consigue, por ejemplo, simular clics y rellenar formularios automáticamente. Los cambios en la web que resultan de dichas acciones se reflejan en el DOM. La estructura del proceso de web scraping con Selenium es la siguiente:

URL → Solicitud HTTP → HTML → Selenium → DOM

### IMDB

```python
def extraer_imdb_250():
    
    options = webdriver.ChromeOptions()
    
    # ejecución en segundo plano (sin abrir la ventana):
    # options.add_argument("--headless")
    
    driver = uc.Chrome()
    datos_peliculas = []
    
    try:
        print("Accediendo a IMDb Top 250...")
        driver.get("https://www.imdb.com/chart/top/")
        
        # 2. Esperamos hasta 15 segundos a que la lista de películas cargue en la pantalla
        wait = WebDriverWait(driver, 15)
        lista_peliculas = wait.until(
            EC.presence_of_all_elements_located((By.CSS_SELECTOR, "li.ipc-metadata-list-summary-item"))
        )
        
        print(f"¡Se encontraron {len(lista_peliculas)} películas! Extrayendo datos...")
        
        # 3. Recorremos cada película para extraer la información
        for pelicula in lista_peliculas:
            try:
                # TÍTULO: Suele venir en formato "1. The Shawshank Redemption"
                titulo_crudo = pelicula.find_element(By.CSS_SELECTOR, "h3.ipc-title__text").text
                # Limpiamos el número inicial (ej: "1. ")
                titulo = titulo_crudo.split(".", 1)[-1].strip() if "." in titulo_crudo else titulo_crudo
                print(titulo)
                # DURACIÓN: IMDb guarda "Año", "Duración" y "Clasificación" en spans idénticos.
                # La duración siempre es el segundo elemento (índice 1).
                metadatos = pelicula.find_elements(By.CSS_SELECTOR, "span.cli-title-metadata-item")
                duracion = metadatos[1].text if len(metadatos) > 1 else "N/A"
                
                # PUNTUACIÓN: Seleccionamos el elemento de la estrella
                puntuacion = pelicula.find_element(By.CSS_SELECTOR, "span.ipc-rating-star--rating").text
                
                # Guardamos en nuestra lista
                datos_peliculas.append({
                    "Título": titulo,
                    "Duración": duracion,
                    "Puntuación": puntuacion
                })
                
            except Exception as e:
                # Si una película puntual tiene un fallo de formato, la saltamos y seguimos
                continue
                
    finally:
        # 4. Cerramos el navegador siempre
        driver.quit()
        
    # 5. Convertimos los datos en un DataFrame y los exportamos
    df = pd.DataFrame(datos_peliculas)
    df.to_csv("imdb_top_250.csv", index=False, encoding="utf-8-sig")
    
    print("\nProceso terminado. Se ha creado el archivo 'imdb_top_250.csv'")
    return df
```
