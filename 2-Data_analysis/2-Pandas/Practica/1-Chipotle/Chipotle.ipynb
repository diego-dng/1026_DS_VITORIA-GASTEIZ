### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
chipo = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep = "\t")
chipo
```

```python
chipo.info()
```

```python
chipo.describe(include="all")
```

### Step 3. Name of the max valued product

```python
chipo["item_price"].max()
```

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "").astype(float)
```

```python
chipo[chipo["item_price"] == chipo["item_price"].max()]
```

```python
chipo["unit_price"] = chipo["item_price"] / chipo["quantity"]
chipo.head()
```

```python
chipo_filtrado = chipo[chipo["unit_price"] == chipo["unit_price"].max()]
chipo_filtrado["item_name"].unique()
```

### Step 4. How many products cost more than $10.00?

```python
chipo[chipo["unit_price"]>10]["item_name"].nunique()
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
chipo.head()
```

```python
chipo[chipo["unit_price"] > 10]["order_id"].nunique()
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 20$? Es lo mismo?

```python
chipo_2 = chipo.groupby("order_id")["item_price"].sum()
len(chipo_2[chipo_2 > 20])
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
chipo[chipo["item_price"] > 10]["order_id"].nunique()
```

```python
chipo[chipo["unit_price"] > 10]["order_id"].nunique()
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
chipo.loc[:, ["item_name", "unit_price"]]
```

```python
chipo[["item_name", "unit_price"]].drop_duplicates()
```

### Step 6. Sort by the name of the item

```python
chipo[["item_name", "unit_price"]].drop_duplicates().sort_values(["item_name"])
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
chipo[chipo["item_price"] == chipo["item_price"].max()]
```

```python
chipo[chipo["unit_price"] == chipo["unit_price"].max()]["quantity"].sum()
```

```python
chipo[chipo["unit_price"] == chipo["unit_price"].max()].groupby("item_name")["quantity"].sum()
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo[chipo["item_name"] == "Veggie Salad Bowl"]["order_id"].nunique()
```

```python
chipo[chipo["item_name"] == "Veggie Salad Bowl"]["quantity"].sum()
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
chipo[(chipo["item_name"] ==  "Canned Soda") & (chipo["quantity"] > 1)]["order_id"].nunique()
```
